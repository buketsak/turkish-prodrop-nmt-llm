```python
nlp = spacy.load("en_core_web_sm")
```

```python
PERSONAL_PRONOUNS = ["i", "you", "he", "she", "we", "they", "it"]


# Function to get subject pronouns
def get_subject_pronouns(text):
    # Expand contractions on the fly
    text = contractions.fix(text)
    doc = nlp(text)
    pronouns = []
    for token in doc:
        if token.dep_ in ("nsubj") and token.pos_ == "PRON":
            if token.text.lower() in PERSONAL_PRONOUNS:
                pronouns.append(token.text.lower())
    return pronouns


# Then fill subject pronouns using the new function
df3["subject_pronouns"] = df3["source_sentence"].apply(get_subject_pronouns)
df2["subject_pronouns"] = df2["source_sentence"].apply(get_subject_pronouns)

# save again after extracting subject pronouns
df3.to_csv('df3_subject_pronouns.csv', index=False) #have few mistakes, must be corrected
df2.to_csv('df2_subject_pronouns.csv', index=False)
```

```python
#df3
import json
def str_to_list_safe(x):
    # If it's already a list, return as is
    if isinstance(x, list):
        return x
    # If it's NaN or empty
    if pd.isna(x) or x in ('', '[]'):
        return []
    # Otherwise, try to parse string
    try:
        # Replace single quotes with double quotes
        x_fixed = str(x).replace("'", '"')
        return json.loads(x_fixed)
    except:
        return []

for df in [df2, df3]:
    df['subject_pronouns'] = df['subject_pronouns'].apply(str_to_list_safe)
    df['pronoun_labels'] = df['pronoun_labels'].apply(str_to_list_safe)
```

```python
df2_sp_done = df2.sort_values(["group_id", "sentence_id"])

df2_sp_done["sentence_index"] = df2_sp_done.groupby("group_id").cumcount() + 1
df2_sp_done["coreference"] = df2_sp_done["sentence_index"].apply(
    lambda x: 1 if x == 1 else 0
)
```

```python
def make_long(df):
    rows = []
    for _, row in df.iterrows():
        if not row["subject_pronouns"] or not row["pronoun_labels"]:
            continue
        for pronoun, overt in zip(row["subject_pronouns"], row["pronoun_labels"]):
            rows.append({
                "group_id": row["group_id"],
                "sentence_id": row["sentence_id"],
                "coreference": row["coreference"],
                "pronoun": pronoun,
                "overt": overt
            })
    return pd.DataFrame(rows)

df2_long = make_long(df2_sp_done)
df2_long.to_csv("df2_long.csv", index=False)
```
